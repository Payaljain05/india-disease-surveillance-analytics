```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
epiclem = pd.read_csv("C:/Users/Rachit/OneDrive/Documents/disease_surveillance/data/processed/epiclem.csv")
```

```python
# Top states with highest number of cases

state_cases = epiclem.groupby('state_ut')['Cases_Per_Lakh'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=state_cases.values, y=state_cases.index)
#index is 
plt.title('Top 10 States by Average Disease Burden (Cases Per Lakh)')
plt.xlabel('Avg Cases per Lakh Population')
plt.ylabel('States')
plt.tight_layout()
# plt.savefig('top10_states.png')
```

```python
# Disease Burden by Season (Avg Cases Per Lakh)

seasonal_pivot = epiclem.pivot_table(
    values='Cases_Per_Lakh',
    index= 'Disease_Grouped',
    columns='Season',
    aggfunc='mean')

plt.figure(figsize=(10,6))
sns.heatmap(seasonal_pivot,annot=True,fmt=".2f",cmap='Blues')
plt.title('Disease Burden by Season (Avg Cases Per Lakh)')
plt.xlabel('Season')
plt.ylabel('Disease Group')
plt.tight_layout()
plt.savefig('seasonal_disease_burden.png')
```

```python
# Distribution of Cases per Lakh
plt.figure(figsize=(10, 5))
sns.histplot(epiclem['Cases_Per_Lakh'], bins=50, kde=True)
plt.title('Distribution of Cases per Lakh Population')
plt.xlabel('Cases per Lakh Population')
plt.ylabel('Number of Outbreaks')
plt.xlim(0, 20) 
plt.tight_layout()
plt.savefig('distribution.png', dpi=150)
plt.show()
```
